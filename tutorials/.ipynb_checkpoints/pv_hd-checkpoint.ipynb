# Coupling PV Production and Heat Demand in `pandaprosumer`

This tutorial demonstrates how to couple a **PV production controller** with a **heat demand / heat pump controller** using the `pandaprosumer` framework.

The idea is similar to the `chiller_demand` tutorial:

- One controller provides **time-series PV production** (e.g. from PVGIS via `pvlib`).
- Another controller provides **heat demand** and controls a **heat pump**.
- Both controllers interact **via the shared `prosumer` object** and **time-series data source**, not by calling each other directly.

You can adapt this notebook to your own models, components and data.


## 1. Environment Setup

Make sure you have the development version of `pandaprosumer` installed (or adjust the import paths to your local clone).

```python
# If needed, install pandaprosumer (uncomment this if you run in a clean environment)
# !pip install pandaprosumer

import numpy as np
import pandas as pd

import pandaprosumer as ppros  # adjust if your import path is different
from pandaprosumer.controller.base import BasicProsumerController

print(ppros.__version__)
```

## 2. Time-Series Data: PV and Heat Demand

In this tutorial we assume you already have time-series data for:

- **PV production** and related variables (power, irradiance, temperature, etc.), e.g. created with a `PvProduction` class calling PVGIS via `pvlib`.
- **Heat demand** (e.g. building heat load).

Here we create simple example profiles to keep the tutorial self-contained. Replace this section with your real PV/heat demand data loading.

```python
# Example time index: one day with hourly resolution
time_index = pd.date_range("2024-01-01", periods=24, freq="H")

# Example PV profile: simple sine-like shape (0 at night, peak at midday)
hours = np.arange(24)
pv_power_w = np.maximum(0, 1000 * np.sin((hours - 6) / 24 * 2 * np.pi))  # 1 kW peak

# Dummy irradiance and weather variables (normally from PVGIS / pvlib)
poa_direct = pv_power_w * 0.7
poa_sky_diffuse = pv_power_w * 0.2
poa_ground_diffuse = pv_power_w * 0.1
solar_elevation_deg = np.clip(60 * np.sin((hours - 6) / 24 * 2 * np.pi), 0, None)
temp_air_c = 10 + 5 * np.sin((hours - 6) / 24 * 2 * np.pi)
wind_speed_m_s = 2 + np.zeros_like(hours)
solar_rad_reconstr_bool = np.ones_like(hours, dtype=int)

pv_df = pd.DataFrame(
    index=time_index,
    data={
        "p_w": pv_power_w,
        "poa_direct_w_m2": poa_direct,
        "poa_sky_diffuse_w_m2": poa_sky_diffuse,
        "poa_ground_diffuse_w_m2": poa_ground_diffuse,
        "solar_elevation_deg": solar_elevation_deg,
        "temp_air_c": temp_air_c,
        "wind_speed_m_s": wind_speed_m_s,
        "solar_rad_reconstr_bool": solar_rad_reconstr_bool,
    },
)

# Example heat demand: higher in morning and evening
heat_demand_kw = 5 + 3 * np.sin((hours - 8) / 24 * 2 * np.pi)  # dummy profile
heat_df = pd.DataFrame(index=time_index, data={"heat_demand_kw": heat_demand_kw})

pv_df.head(), heat_df.head()
```

## 3. Build a Simple Prosumer and DataSource

In your real project you may already have a `Prosumer` object and a dedicated `DataSource` implementation.

Here, we show a minimal sketch where the data source wraps the two dataframes `pv_df` and `heat_df`.


```python
class SimpleDataSource:
    """Minimal data source to serve time-series values.

    This is a simplified stand-in for pandaprosumer's more flexible data sources.
    It supports keys like (element, column) to distinguish between PV and heat demand.
    """

    def __init__(self, pv_df, heat_df):
        self.pv_df = pv_df
        self.heat_df = heat_df

    def get_time_step_value(self, time, key, element=None, index=None):
        """Return the value for a given time and key.

        Parameters
        ----------
        time : timestamp
            Current time step
        key : str
            Column name (e.g. "p_w", "heat_demand_kw")
        element : str
            Element name (e.g. "sn_pv_production" or "sn_heat_demand")
        index : list or None
            Element indices (not used in this simple example, assuming single element)
        """
        if element == "sn_pv_production":
            return self.pv_df.loc[time, key]
        elif element == "sn_heat_demand":
            return self.heat_df.loc[time, key]
        else:
            raise KeyError(f"Unknown element {element}")


data_source = SimpleDataSource(pv_df, heat_df)
```

Next, we create a very simple `prosumer` structure with two element tables:

- `sn_pv_production` – holding PV-related data
- `sn_heat_demand` – holding heat demand and results for the heat pump


```python
# Minimal prosumer-like container using dict of DataFrames
prosumer = {}

# PV element table: one row (index 0)
prosumer["sn_pv_production"] = pd.DataFrame(index=[0])

# Heat demand element table: one row (index 0)
prosumer["sn_heat_demand"] = pd.DataFrame(index=[0])

prosumer
```

## 4. PV Component Data

We define a component data class for the PV production, similarly to your
`SenergyNetsPvProductionComponentData`. This carries the list of element indices,
input columns and result columns.

```python
from dataclasses import dataclass, field
from typing import List


@dataclass
class SenergyNetsPvProductionComponentData:
    element_index: List[int]
    input_columns: List[str] = field(
        default_factory=lambda: [
            "p_w",
            "poa_direct_w_m2",
            "poa_sky_diffuse_w_m2",
            "poa_ground_diffuse_w_m2",
            "solar_elevation_deg",
            "temp_air_c",
            "wind_speed_m_s",
            "solar_rad_reconstr_bool",
        ]
    )
    result_columns: List[str] = field(
        default_factory=lambda: [
            "p_w",
            "poa_direct_w_m2",
            "poa_sky_diffuse_w_m2",
            "poa_ground_diffuse_w_m2",
            "solar_elevation_deg",
            "temp_air_c",
            "wind_speed_m_s",
            "solar_rad_reconstr_bool",
        ]
    )
    period_index: int = None
    location_index: int = None
    element: str = "sn_pv_production"
    name: str = "sn_pv_production"


pv_comp = SenergyNetsPvProductionComponentData(element_index=[0])
pv_comp
```

## 5. Implement the PV Production Controller

Here is a simplified version of your `SenergyNetsPvProductionController`.

- It reads PV-related time-series values from the `data_source` in `time_step`.
- It writes them into `self.res` and also into the `prosumer["sn_pv_production"]` table.
- A heat demand controller can later read `p_w` from that table.


```python
class SenergyNetsPvProductionController(BasicProsumerController):
    @classmethod
    def name(cls):
        return "sn_pvproduction"

    def __init__(self, prosumer, pv_production_object, order, level, data_source,
                 in_service=True, index=None, name=None, **kwargs):
        super().__init__(
            prosumer,
            pv_production_object,
            order=order,
            level=level,
            data_source=data_source,
            in_service=in_service,
            index=index,
            name=name,
            **kwargs,
        )

        self.element = self.obj.element
        self.element_index = self.obj.element_index
        self.result_columns = self.obj.result_columns
        self.pros = prosumer[self.element].loc[self.element_index, :]
        self.res = np.zeros(
            [len(self.element_index), len(self.time_index), len(self.result_columns)]
        )

        self.res_step = None
        self.time = None
        self.applied = None

    def time_step(self, prosumer, time):
        super().time_step(prosumer, time)
        self.res_step = np.zeros(
            [len(self.element_index), len(self.obj.result_columns)]
        )
        self.time = time
        self.applied = False

        # Read all PV-related inputs from data source
        self._p_w = self.ds.get_time_step_value(time, "p_w", element=self.element)
        self._poa_direct_w_m2 = self.ds.get_time_step_value(time, "poa_direct_w_m2", element=self.element)
        self._poa_sky_diffuse_w_m2 = self.ds.get_time_step_value(time, "poa_sky_diffuse_w_m2", element=self.element)
        self._poa_ground_diffuse_w_m2 = self.ds.get_time_step_value(time, "poa_ground_diffuse_w_m2", element=self.element)
        self._solar_elevation_deg = self.ds.get_time_step_value(time, "solar_elevation_deg", element=self.element)
        self._temp_air_c = self.ds.get_time_step_value(time, "temp_air_c", element=self.element)
        self._wind_speed_m_s = self.ds.get_time_step_value(time, "wind_speed_m_s", element=self.element)
        self._solar_rad_reconstr_bool = self.ds.get_time_step_value(time, "solar_rad_reconstr_bool", element=self.element)

    def is_converged(self, container):
        return self.applied

    def control_step(self, prosumer):
        super().control_step(prosumer)

        # Bundle results in the same order as result_columns
        result = (
            self._p_w,
            self._poa_direct_w_m2,
            self._poa_sky_diffuse_w_m2,
            self._poa_ground_diffuse_w_m2,
            self._solar_elevation_deg,
            self._temp_air_c,
            self._wind_speed_m_s,
            self._solar_rad_reconstr_bool,
        )

        # Write PV power (and optionally other variables) into the prosumer table
        # Here we assume one element index [0]
        idx = self.element_index[0]
        prosumer[self.element].loc[idx, "p_w"] = self._p_w

        # Store result in res via the base class finalize method (if available),
        # or write into self.res manually
        t_idx = np.where(self.time_index == self.time)[0][0]
        self.res[0, t_idx, :] = np.array(result)

        self.applied = True

    def time_series_finalization(self, prosumer):
        return self.res


# Instantiate PV controller
pv_controller = SenergyNetsPvProductionController(
    prosumer=prosumer,
    pv_production_object=pv_comp,
    order=["sn_pv_production"],
    level=[0],
    data_source=data_source,
    in_service=True,
    index=None,
    name="PV Controller",
)
pv_controller
```

## 6. Implement a Simple Heat Demand / Heat Pump Controller

Now we implement a controller that:

- Reads **heat demand** time series.
- Reads **PV electrical power** from `prosumer["sn_pv_production"]`.
- Assumes a simple heat pump with a fixed COP (coefficient of performance).
- Prioritises using PV electricity; any remaining electric demand is assumed to be met by the grid.


```python
@dataclass
class HeatDemandComponentData:
    element_index: List[int]
    input_columns: List[str] = field(default_factory=lambda: ["heat_demand_kw"])
    result_columns: List[str] = field(
        default_factory=lambda: ["heat_pump_elec_kw", "grid_elec_kw"]
    )
    element: str = "sn_heat_demand"
    name: str = "sn_heat_demand"


heat_comp = HeatDemandComponentData(element_index=[0])
heat_comp
```

```python
class SimpleHeatDemandController(BasicProsumerController):
    @classmethod
    def name(cls):
        return "sn_heat_demand_ctrl"

    def __init__(self, prosumer, component_object, order, level, data_source,
                 in_service=True, index=None, name=None, cop=3.0, **kwargs):
        super().__init__(
            prosumer,
            component_object,
            order=order,
            level=level,
            data_source=data_source,
            in_service=in_service,
            index=index,
            name=name,
            **kwargs,
        )
        self.element = self.obj.element
        self.element_index = self.obj.element_index
        self.result_columns = self.obj.result_columns
        self.pros = prosumer[self.element].loc[self.element_index, :]
        self.res = np.zeros(
            [len(self.element_index), len(self.time_index), len(self.result_columns)]
        )
        self.cop = cop  # coefficient of performance of the heat pump
        self.time = None
        self.applied = None

    def time_step(self, prosumer, time):
        super().time_step(prosumer, time)
        self.time = time
        self.applied = False
        # Read heat demand
        self._heat_demand_kw = self.ds.get_time_step_value(
            time, "heat_demand_kw", element=self.element
        )

    def is_converged(self, container):
        return self.applied

    def control_step(self, prosumer):
        super().control_step(prosumer)

        idx_heat = self.element_index[0]
        idx_pv = 0  # PV element index

        # Heat pump electrical power needed (kW)
        hp_elec_kw_needed = self._heat_demand_kw / self.cop

        # Available PV power (convert from W to kW)
        pv_p_w = prosumer["sn_pv_production"].loc[idx_pv, "p_w"]
        pv_p_kw = pv_p_w / 1000.0

        # Use PV as much as possible
        hp_elec_kw_from_pv = min(hp_elec_kw_needed, pv_p_kw)
        grid_elec_kw = hp_elec_kw_needed - hp_elec_kw_from_pv

        # Store into prosumer table
        prosumer[self.element].loc[idx_heat, "heat_pump_elec_kw"] = hp_elec_kw_from_pv
        prosumer[self.element].loc[idx_heat, "grid_elec_kw"] = grid_elec_kw

        # Store into result array
        t_idx = np.where(self.time_index == self.time)[0][0]
        self.res[0, t_idx, :] = np.array([hp_elec_kw_from_pv, grid_elec_kw])

        self.applied = True

    def time_series_finalization(self, prosumer):
        return self.res


heat_controller = SimpleHeatDemandController(
    prosumer=prosumer,
    component_object=heat_comp,
    order=["sn_heat_demand"],
    level=[0],
    data_source=data_source,
    in_service=True,
    index=None,
    name="Heat Demand Controller",
    cop=3.0,
)
heat_controller
```

## 7. Run a Simple Time-Series Loop

In the full pandaprosumer framework, you would typically use a dedicated
time-series simulation engine and `ControlHandler`. Here we show a minimal
manual loop over time to keep the tutorial self-contained.


```python
pv_results = []
heat_results = []

for t in time_index:
    # PV controller step
    pv_controller.time_step(prosumer, t)
    pv_controller.control_step(prosumer)

    # Heat controller step (uses PV power written by PV controller)
    heat_controller.time_step(prosumer, t)
    heat_controller.control_step(prosumer)

pv_res = pv_controller.time_series_finalization(prosumer)[0]
heat_res = heat_controller.time_series_finalization(prosumer)[0]

pv_res.shape, heat_res.shape
```

## 8. Plot Results

We plot PV production, heat demand and how much of the heat pump electricity
is supplied by PV versus the grid.


```python
import matplotlib.pyplot as plt

# Extract columns based on result_columns
pv_cols = pv_comp.result_columns
pv_p_w_series = pv_res[:, pv_cols.index("p_w")]

heat_cols = heat_comp.result_columns
hp_elec_kw_series = heat_res[:, heat_cols.index("heat_pump_elec_kw")]
grid_elec_kw_series = heat_res[:, heat_cols.index("grid_elec_kw")]

plt.figure(figsize=(10, 6))
plt.plot(time_index, pv_p_w_series / 1000.0, label="PV power [kW]")
plt.plot(time_index, heat_df["heat_demand_kw"], label="Heat demand [kW]")
plt.plot(time_index, hp_elec_kw_series, label="Heat pump elec from PV [kW]")
plt.plot(time_index, grid_elec_kw_series, label="Grid elec for heat pump [kW]")
plt.legend()
plt.grid(True)
plt.xlabel("Time")
plt.ylabel("Power [kW]")
plt.title("Coupled PV and Heat Demand")
plt.tight_layout()
plt.show()
```

## 9. Discussion and Next Steps

- The PV controller provides **PV power and related variables** to the prosumer model.
- The heat demand controller reads **heat demand** and **PV power** and decides how much
  of the heat pump's electric consumption can be covered by PV.
- In a full pandaprosumer setup, you would integrate these controllers into the
  `ControlHandler` and the standard time-series simulation tools.

Possible extensions:

- Add **thermal storage** (buffer tank) to shift heat demand.
- Add **electric storage** (battery) to increase PV self-consumption.
- Include **electricity prices** and optimise operation economically.
